# Nightly H10 ECG and two PPG recordings

One Qt5Agg window per diary-defined night, with shared local-time axes for H10 ECG, Verity Sense PPG, and EmbracePlus PPG. Values remain in their exported units; no filtering, normalization, resampling, gap filling, or clock correction is applied. Missing recordings are labeled. Overview drawing retains extrema, and zooming redraws the original samples.

Use an environment with `offline_processing/requirements-notebooks.txt` installed. Set `night_data_export_path` in the ignored `paths.local.json`, or set `OVERNIGHT_NIGHT_DATA_EXPORT_PATH` to your export directory.

```python
import matplotlib as mpl
mpl.use("Qt5Agg")
from IPython import get_ipython
get_ipython().run_line_magic("matplotlib", "qt5")
import matplotlib.pyplot as plt
import pandas as pd

from notebook_paths import input_path
from nightly_signals import load_nights, plot_night

export_directory = input_path("night_data_export_path")
nights = load_nights(export_directory)
```

```python
summary = pd.DataFrame([
    {"night": night.diary_date, "start": night.start, "end": night.end,
     **{key + "_samples": len(signal) for key, signal in night.signals.items()}}
    for night in nights
]).set_index("night")
summary
```

```python
plt.close("all")
figures = [plot_night(night) for night in nights]
plt.show(block=False)
```
